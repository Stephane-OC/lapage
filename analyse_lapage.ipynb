# 🎯 Objectif du projet — Lapage 📚

Ce notebook a pour objectif d’**analyser les ventes et le comportement des clients de l’entreprise Lapage**, une librairie ayant développé son activité de vente en ligne, afin d’en extraire des **insights clairs et exploitables pour le comité de direction (CODIR)**.

L’analyse vise à fournir une **vision synthétique et stratégique de la performance commerciale**, ainsi qu’une meilleure compréhension des profils et habitudes des clients.

---

## 🧩 Contexte

Initialement positionnée comme une **librairie physique**, l’entreprise **Lapage** a lancé son site de vente en ligne il y a deux ans afin de répondre à la demande croissante de ses clients et d’élargir son activité.

Dans ce contexte, la direction souhaite aujourd’hui :

- mieux **suivre l’évolution de l’activité e-commerce** ;
- comprendre les **dynamiques de vente** ;
- analyser le **comportement des clients en ligne**.

Pour mener cette étude, trois jeux de données sont mis à disposition :

- un fichier **clients** (genre, année de naissance) ;
- un fichier **produits** (prix et catégories) ;
- un fichier **transactions** (dates d’achat, clients et produits).

---

## 📈 Objectifs principaux

Le travail se structure autour de **deux axes complémentaires**.

### Axe 1 — Analyse de l’activité commerciale

- Étudier l’**évolution du chiffre d’affaires dans le temps** ;
- Mettre en évidence les tendances à l’aide de **moyennes mobiles** ;
- Analyser le **nombre de clients**, de **transactions** et de **produits vendus** ;
- Identifier les **produits les plus performants (tops)** et les **moins performants (flops)** ;
- Étudier la **répartition des ventes par catégorie**.

---

### Axe 2 — Analyse du comportement client

- Étudier la **répartition du chiffre d’affaires par client** ;
- Analyser la **dispersion des ventes** à l’aide d’une **courbe de Lorenz** ;
- Mettre en évidence des **corrélations entre les variables clients** :
  - âge ;
  - genre ;
  - fréquence d’achat ;
  - panier moyen ;
  - catégories de livres achetées.
- Appliquer et justifier l’utilisation de **tests statistiques adaptés** :
  - test du **χ² (khi-deux)** ;
  - **ANOVA** ou **test de Student** ;
  - **corrélation de Pearson ou de Spearman**.
- Identifier et traiter les **clients atypiques (BtoB)** afin de garantir la pertinence des analyses.

---

## 🧠 Finalité

Les résultats issus de cette étude permettront :

- d’**accompagner la prise de décision stratégique du CODIR** ;
- d’orienter les **choix marketing et commerciaux** (offres, ciblage, positionnement) ;
- de proposer une **lecture claire, pédagogique et exploitable des données**, adaptée à un public non technique.

---

> 💡 *Cette analyse a été menée dans une démarche de rigueur, de clarté et de reproductibilité, afin de fournir à Lapage une vision fiable et structurée de son activité e-commerce et du comportement de ses clients.*

```python
cols_tx = ["id_prod", "date", "session_id", "client_id"]
rows = len(transactions)
nulls = transactions[cols_tx].isna().sum()

summary_missing = pd.DataFrame({
    "Non-null": rows - nulls,
    "Valeurs manquantes": nulls,
    "% manquantes": (nulls / rows * 100).round(2)
})

display(summary_missing)

# Lignes "exploitables" = toutes les colonnes clés présentes
tx_ok = transactions.dropna(subset=cols_tx).copy()

# Lignes "incomplètes" = au moins 1 colonne clé manquante
tx_missing = transactions[transactions[cols_tx].isna().any(axis=1)].copy()

# Cas très fréquent : lignes "totalement vides" (toutes NaN sur les 4 colonnes)
tx_all_nan = transactions[transactions[cols_tx].isna().all(axis=1)].copy()

print("Nb lignes total :", rows)
print("Nb lignes exploitables (4 champs présents) :", len(tx_ok))
print("Nb lignes incomplètes (>=1 champ manquant) :", len(tx_missing))
print("Nb lignes entièrement vides (4 champs manquants) :", len(tx_all_nan))
```

```python
# Contrôle 1 — Détection des doublons techniques (Transactions exploitables)

dup_exact_ok = tx_ok.duplicated().sum()
dup_strict_ok = tx_ok.duplicated(subset=cols_tx).sum()

print("\nDoublons exacts (ligne complète) sur lignes exploitables :", dup_exact_ok)
print("Doublons stricts sur (client, session, produit, date) sur lignes exploitables :", dup_strict_ok)

if dup_exact_ok > 0:
    display(tx_ok[tx_ok.duplicated(keep=False)].sort_values(cols_tx).head(20))
```

```python
# Contrôle 2 — Répétitions métier (achats multiples dans une même session)

tx_repeat = (
    tx_ok.groupby(["client_id", "session_id", "id_prod"])
         .size()
         .reset_index(name="nb_occurrences")
)

triplets = len(tx_repeat)
rep_gt1 = (tx_repeat["nb_occurrences"] > 1).sum()
part_rep = (rep_gt1 / triplets * 100) if triplets else 0
max_rep = tx_repeat["nb_occurrences"].max() if triplets else 0

print("\nTriplets analysés (client, session, produit) :", triplets)
print("Répétitions (nb_occurrences > 1) :", rep_gt1)
print("Part des répétitions : {:.2f}%".format(part_rep))
print("Maximum observé (même produit, même session) :", max_rep)

# Aperçu des cas les + élevés
display(tx_repeat.sort_values("nb_occurrences", ascending=False).head(10))
```

```python
# Préparer les données de CA (contrôle des colonnes et de la période)
df = dataset_final.copy()

# Vérifications rapides
print("Colonnes dispo :", df.columns.tolist())
print("Shape :", df.shape)

# Sécuriser les types
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["price"] = pd.to_numeric(df["price"], errors="coerce")

# Contrôle des NA
print("\nNA date :", df["date"].isna().sum())
print("NA price :", df["price"].isna().sum())

# Période couverte
print("\nPériode :", df["date"].min(), "→", df["date"].max())

# On garde uniquement les lignes exploitables pour le CA
df_ca = df.dropna(subset=["date", "price"]).copy()

# CA négatif ? (normalement non)
print("\nPrix < 0 :", (df_ca["price"] < 0).sum())
print("Prix = 0 :", (df_ca["price"] == 0).sum())
```

```python
# Calcul du chiffre d'affaires par jour + tendance (moyenne mobile 7j)
ca_daily = (
    df_ca
    .groupby(df_ca["date"].dt.date)["price"]
    .sum()
    .rename("ca")
    .to_frame()
)

ca_daily.index = pd.to_datetime(ca_daily.index)
ca_daily = ca_daily.sort_index()

# Ajout moyenne mobile 7 jours (tendance)
ca_daily["ca_mm7"] = ca_daily["ca"].rolling(7, min_periods=1).mean()

display(ca_daily.head())
display(ca_daily.tail())

print("Nb jours :", len(ca_daily))
print("CA total :", round(ca_daily["ca"].sum(), 2))
print("CA moyen / jour :", round(ca_daily["ca"].mean(), 2))
print("Jour max :", ca_daily["ca"].idxmax().date(), "=", round(ca_daily["ca"].max(), 2))
```

```python
# Graphique CA journalier + moyenne mobile (7 jours)
plt.figure(figsize=(12, 5), facecolor="#0f111a")
ax = plt.gca()
ax.set_facecolor("#0f111a")

ax.plot(ca_daily.index, ca_daily["ca"], linewidth=1, label="Daily revenue")
ax.plot(ca_daily.index, ca_daily["ca_mm7"], linewidth=2, label="7-day moving average")

ax.set_title("Revenue over time (daily)", color="white", pad=12)
ax.set_xlabel("Date", color="white")
ax.set_ylabel("Revenue", color="white")

ax.tick_params(colors="white")
ax.grid(axis="y", alpha=0.20)

leg = ax.legend()
for text in leg.get_texts():
    text.set_color("white")

plt.tight_layout()
plt.show()
```

```python
# Regarder la catégorie qui a eu une baisse le chiffre d'affaire et l'expliquer
# CA mensuel par catégorie
df_ca["month"] = df_ca["date"].dt.to_period("M").dt.to_timestamp()

ca_cat_month = (
    df_ca
    .groupby(["month", "categ"])["price"]
    .sum()
    .rename("ca")
    .reset_index()
    .sort_values(["categ", "month"])
)

display(ca_cat_month.head())
```

```python
# Variation mensuelle (%) par catégorie
ca_cat_month["ca_prev"] = (
    ca_cat_month
    .groupby("categ")["ca"]
    .shift(1)
)

ca_cat_month["ca_change_pct"] = (
    (ca_cat_month["ca"] - ca_cat_month["ca_prev"])
    / ca_cat_month["ca_prev"] * 100
).round(2)

display(ca_cat_month.head())
```

```python
# Détection de la plus forte baisse mensuelle du chiffre d'affaires

# On supprime les lignes sans variation calculée
ca_cat_valid = ca_cat_month.dropna(subset=["ca_change_pct"]).copy()

# Identification de la ligne correspondant à la baisse la plus importante
# (valeur minimale du pourcentage de variation)
worst_drop = ca_cat_valid.loc[
    ca_cat_valid["ca_change_pct"].idxmin()
]

print("Plus forte baisse observée :")
display(worst_drop)
```

```python
# Volume de transactions par catégorie et par mois
tx_cat_month = (
    df_ca
    .groupby(["month", "categ"])
    .size()
    .rename("transactions")
    .reset_index()
)

# Fusion avec ton CA catégorie
ca_cat_full = ca_cat_month.merge(
    tx_cat_month,
    on=["month", "categ"],
    how="left"
)

display(
    ca_cat_full[
        (ca_cat_full["categ"] == worst_drop["categ"])
        & (ca_cat_full["month"].isin([
            worst_drop["month"],
            worst_drop["month"] - pd.offsets.MonthBegin(1)
        ]))
    ]
)
```

```python
# Analyse complémentaire : panier moyen autour de la baisse

# Calcul du panier moyen (CA / nombre de transactions)
ca_cat_full["panier_moyen"] = (
    ca_cat_full["ca"] / ca_cat_full["transactions"]
).round(2)

# Affichage de la catégorie concernée
# pour le mois de la baisse et le mois précédent
display(
    ca_cat_full[
        (ca_cat_full["categ"] == worst_drop["categ"])
        & (ca_cat_full["month"].isin([
            worst_drop["month"],
            worst_drop["month"] - pd.offsets.MonthBegin(1)
        ]))
    ]
)
```

```python
# Visualisation : évolution du chiffre d'affaires
# pour la catégorie ayant subi la plus forte baisse
cat_focus = worst_drop["categ"]

plt.figure(figsize=(10,5))
sns.lineplot(
    data=ca_cat_month[ca_cat_month["categ"] == cat_focus],
    x="month",
    y="ca"
)
plt.title(f"Évolution du CA — Catégorie {cat_focus}")
plt.show()
```

```python
# Préparation des données (périodes + contrôles)
df = dataset_final.copy()

# Sécuriser les types
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["price"] = pd.to_numeric(df["price"], errors="coerce")

# Garder uniquement les lignes exploitables pour l'analyse par période
df_period = df.dropna(subset=["date", "price", "client_id", "id_prod"]).copy()

# Colonnes de période (jour / semaine / mois)
df_period["day"] = df_period["date"].dt.floor("D")
df_period["week"] = df_period["date"].dt.to_period("W").dt.start_time
df_period["month"] = df_period["date"].dt.to_period("M").dt.start_time

print("✅ Lignes exploitables :", len(df_period))
print("📅 Période :", df_period["day"].min().date(), "→", df_period["day"].max().date())
print("🧾 Transactions :", len(df_period))
print("👤 Clients uniques :", df_period["client_id"].nunique())
print("🛒 Produits uniques :", df_period["id_prod"].nunique())
```

```python
# Agrégation mensuelle (Transactions / Clients / Produits)
kpi_month = (
    df_period.groupby("month")
    .agg(
        transactions=("id_prod", "size"),
        clients_uniques=("client_id", "nunique"),
        produits_uniques=("id_prod", "nunique"),
        ca=("price", "sum"),
    )
    .sort_index()
)

# Petites métriques utiles
kpi_month["ca_par_transaction"] = (kpi_month["ca"] / kpi_month["transactions"]).round(2)
kpi_month["transactions_par_client"] = (kpi_month["transactions"] / kpi_month["clients_uniques"]).round(2)

display(kpi_month.head(6))
display(kpi_month.tail(6))

print("✅ Nombre de mois :", len(kpi_month))
print("📈 Mois max transactions :", kpi_month["transactions"].idxmax().date(), "-", int(kpi_month["transactions"].max()))
print("👥 Mois max clients :", kpi_month["clients_uniques"].idxmax().date(), "-", int(kpi_month["clients_uniques"].max()))
print("🛍️ Mois max produits uniques :", kpi_month["produits_uniques"].idxmax().date(), "-", int(kpi_month["produits_uniques"].max()))
```

```python
# Graphique CA mensuel + Transactions
plt.figure(figsize=(12, 5), facecolor="#0f111a")
ax = plt.gca()
ax.set_facecolor("#0f111a")

ax.plot(kpi_month.index, kpi_month["ca"], linewidth=2.5, label="CA mensuel (€)")
ax.plot(kpi_month.index, kpi_month["transactions"], linewidth=2, alpha=0.8, label="Transactions")

ax.set_title("CA et volume de transactions — par mois", color="white", pad=12)
ax.set_xlabel("Mois", color="white")
ax.set_ylabel("Valeur", color="white")
ax.tick_params(colors="white")
ax.grid(axis="y", alpha=0.20)

leg = ax.legend()
for t in leg.get_texts():
    t.set_color("white")

plt.tight_layout()
plt.show()
```

```python
# Synthèse (valeurs clés)
nb_months = len(kpi_month)

month_max_tx = kpi_month["transactions"].idxmax().date()
max_tx = int(kpi_month["transactions"].max())

month_max_clients = kpi_month["clients_uniques"].idxmax().date()
max_clients = int(kpi_month["clients_uniques"].max())

month_max_prods = kpi_month["produits_uniques"].idxmax().date()
max_prods = int(kpi_month["produits_uniques"].max())

avg_tx_month = int(round(kpi_month["transactions"].mean(), 0))
avg_clients_month = int(round(kpi_month["clients_uniques"].mean(), 0))
avg_prods_month = int(round(kpi_month["produits_uniques"].mean(), 0))

avg_tx_per_client = round(kpi_month["transactions_par_client"].mean(), 2)
avg_ca_per_tx = round(kpi_month["ca_par_transaction"].mean(), 2)

print("📌 Résumé Étape 4.3")
print("- Mois analysés :", nb_months)
print("- Moyenne transactions/mois :", avg_tx_month)
print("- Moyenne clients uniques/mois :", avg_clients_month)
print("- Moyenne produits uniques/mois :", avg_prods_month)
print("- Moyenne transactions/client (mensuel) :", avg_tx_per_client)
print("- Panier moyen (CA/transaction) :", avg_ca_per_tx, "€")
print("- Pic transactions :", month_max_tx, "-", max_tx)
print("- Pic clients :", month_max_clients, "-", max_clients)
print("- Pic produits uniques :", month_max_tx, "-", max_prods)
```

```python
# Concentration du CA
# Calcul du chiffre d'affaires par client
print("Étape 6.2 — Concentration du chiffre d'affaires\n")

# Agrégation CA par client
ca_client = (
    dataset_final
    .groupby("client_id")["price"]
    .sum()
    .reset_index(name="ca_total")
)

display(ca_client.head())

print(f"Nombre total de clients : {len(ca_client)}")
print(f"CA total global : {ca_client['ca_total'].sum():,.0f} €".replace(",", " "))
```

```python
# Analyse de concentration
# Quelle part du CA est générée par les top clients ?

# Tri décroissant
ca_client_sorted = ca_client.sort_values("ca_total", ascending=False).reset_index(drop=True)

total_ca = ca_client_sorted["ca_total"].sum()

# Calcul cumulatif
ca_client_sorted["ca_cumule"] = ca_client_sorted["ca_total"].cumsum()
ca_client_sorted["part_cumulee"] = ca_client_sorted["ca_cumule"] / total_ca

n_clients = len(ca_client_sorted)

top_10 = int(n_clients * 0.10)
top_20 = int(n_clients * 0.20)

part_top_10 = ca_client_sorted.iloc[top_10-1]["part_cumulee"]
part_top_20 = ca_client_sorted.iloc[top_20-1]["part_cumulee"]

print(f"Top 10% des clients génèrent {part_top_10*100:.1f}% du CA")
print(f"Top 20% des clients génèrent {part_top_20*100:.1f}% du CA")
```

```python
# Graphique Courbe de Lorenz

plt.style.use("dark_background")

# Tri croissant pour Lorenz
ca_sorted_asc = np.sort(ca_client["ca_total"].values)

# Cumuls
ca_cumsum = np.cumsum(ca_sorted_asc)
ca_cumsum = np.insert(ca_cumsum, 0, 0)
ca_cumsum = ca_cumsum / ca_cumsum[-1]

clients_cum = np.linspace(0, 1, len(ca_cumsum))

fig, ax = plt.subplots(figsize=(7,5))

ax.plot(clients_cum, ca_cumsum, linewidth=2)
ax.plot([0,1], [0,1], linestyle="--")

ax.set_title("Courbe de Lorenz — Concentration du CA", pad=15)
ax.set_xlabel("Part cumulée des clients")
ax.set_ylabel("Part cumulée du CA")
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
# Coefficient de Gini

# Formule Gini
n = len(ca_sorted_asc)
index = np.arange(1, n+1)

gini = (2 * np.sum(index * ca_sorted_asc)) / (n * np.sum(ca_sorted_asc)) - (n + 1) / n

print(f"Coefficient de Gini : {gini:.3f}")

if gini < 0.3:
    print("➡️ Concentration faible")
elif gini < 0.5:
    print("➡️ Concentration modérée")
else:
    print("➡️ Concentration forte (effet Pareto marqué)")
```

```python
# Objectif : repérer des clients présentant des volumes/CA très élevés
# souvent associés à des achats d'entreprise (BtoB) pouvant biaiser
# les analyses statistiques
CLIENT_COL = "client_id"

clients_kpi = (
    dataset_final
    .groupby(CLIENT_COL)
    .agg(
        nb_transactions=("id_prod", "count"),
        quantite_totale=("quantity", "sum") if "quantity" in dataset_final.columns else ("id_prod", "count"),
        ca_total=("price", "sum")
    )
    .reset_index()
)

display(clients_kpi.sort_values("ca_total", ascending=False).head(10))
```

```python
# Détection des outliers via IQR (robuste)
# Critères : CA total et/ou nb_transactions très élevés
def iqr_bounds(series, k=1.5):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    return q1 - k * iqr, q3 + k * iqr

low_ca, high_ca = iqr_bounds(clients_kpi["ca_total"], k=1.5)
low_tx, high_tx = iqr_bounds(clients_kpi["nb_transactions"], k=1.5)

clients_kpi["atypique_ca"] = clients_kpi["ca_total"] > high_ca
clients_kpi["atypique_tx"] = clients_kpi["nb_transactions"] > high_tx

clients_atypiques = clients_kpi[clients_kpi["atypique_ca"] | clients_kpi["atypique_tx"]].copy()

print("Seuil CA (IQR) :", round(high_ca, 2))
print("Seuil transactions (IQR) :", int(high_tx))
print("Nombre de clients atypiques détectés :", len(clients_atypiques))

display(clients_atypiques.sort_values("ca_total", ascending=False).head(20))
```

```python
# Check colonnes — aperçu complet du dataset "sans BtoB"
df = df_sans_btob.copy()

print("Shape :", df.shape)
display(pd.DataFrame({
    "column": df.columns,
    "dtype": [str(df[c].dtype) for c in df.columns],
    "missing_%": [(df[c].isna().mean()*100) for c in df.columns],
    "n_unique": [df[c].nunique(dropna=True) for c in df.columns],
}).sort_values("missing_%", ascending=False))
```

```python
# Colonnes catégorielles candidates (pour χ²)
df = df_sans_btob.copy()

# candidates: object / category
cat_cols = []

for c in df.columns:
    if df[c].dtype == "object" or str(df[c].dtype) == "category":
        cat_cols.append(c)
    else:
        # ex: int avec peu de modalités (0/1/2/3...)
        if pd.api.types.is_integer_dtype(df[c]) and df[c].nunique() <= 20:
            cat_cols.append(c)

print("Colonnes catégorielles candidates :", cat_cols)

# Aperçu des modalités pour les + pertinentes (limité)
for c in cat_cols[:10]:
    print("\n---", c, "---")
    display(df[c].value_counts().head(10))
```

```python
# Focus sur la variable "categ"
if "categ" in df.columns:
    print("Modalités de categ :", df["categ"].unique())
    display(df["categ"].value_counts(normalize=True).mul(100).round(2).rename("percent"))
else:
    print("⚠️ La colonne 'categ' n'existe pas dans df_sans_btob.")
```

```python
# Base — Construire l'âge et un dataset client pour les tests
# Objectif : préparer les variables Age, CA total, fréquence et panier moyen
df = df_sans_btob.copy()

# âge au moment de l'achat (année transaction - année naissance)
df["age"] = df["date"].dt.year - df["birth"]

# sécurité : on garde des âges plausibles (à ajuster si besoin)
df = df[(df["age"] >= 10) & (df["age"] <= 100)].copy()

# dataset client
client_stats = (
    df.groupby("client_id")
      .agg(
          age=("age", "median"),             # âge "représentatif" du client
          nb_transactions=("price", "size"),
          ca_total=("price", "sum"),
          panier_moyen=("price", "mean"),
          sex=("sex", "first")               # 1 valeur par client
      )
      .reset_index()
)

display(client_stats.head())
print("Clients analysés :", len(client_stats))
```

```python
# Test du χ² : Sexe vs Catégorie produit (categ)
# Objectif : vérifier s'il existe une dépendance entre le sexe du client (f/m)
# et la catégorie de produits achetés (categ)
from scipy.stats import chi2_contingency

cont = pd.crosstab(df["sex"], df["categ"])
chi2, p, dof, expected = chi2_contingency(cont)

print(f"χ²(sex vs categ) = {chi2:.2f} | ddl={dof} | p-value={p:.4f}")
display(cont)

# Graph % par sexe
pct = cont.div(cont.sum(axis=1), axis=0) * 100

plt.style.use("dark_background")
fig, ax = plt.subplots(figsize=(8,4))
pct.plot(kind="bar", ax=ax)
ax.set_title("Répartition des catégories (categ) par sexe (%)")
ax.set_xlabel("Sexe")
ax.set_ylabel("Pourcentage (%)")
ax.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# χ² — Tranche d'âge vs catégorie de livres achetés (categ)
# Objectif : vérifier si les catégories varient selon l'âge
from scipy.stats import chi2_contingency

# Tranches d’âge (modifiables)
bins = [10, 20, 30, 40, 50, 60, 100]
labels = ["10-19", "20-29", "30-39", "40-49", "50-59", "60+"]

df["age_group"] = pd.cut(df["age"], bins=bins, labels=labels, right=False)

cont_age = pd.crosstab(df["age_group"], df["categ"])
chi2, p, dof, expected = chi2_contingency(cont_age)

print(f"χ²(age_group vs categ) = {chi2:.2f} | ddl={dof} | p-value={p:.4f}")
display(cont_age)

# Heatmap des résidus standardisés
expected_df = pd.DataFrame(expected, index=cont_age.index, columns=cont_age.columns)
residuals = (cont_age - expected_df) / np.sqrt(expected_df)

plt.style.use("dark_background")
fig, ax = plt.subplots(figsize=(10, 4))
im = ax.imshow(residuals.values, aspect="auto")
ax.set_title("χ² — Résidus standardisés (tranche d'âge vs categ)")
ax.set_xticks(range(len(residuals.columns)))
ax.set_xticklabels(residuals.columns)
ax.set_yticks(range(len(residuals.index)))
ax.set_yticklabels(residuals.index)
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import kstest, probplot, norm, levene

plt.style.use("dark_background")

def ks_normality_report(x, label):
    x = np.asarray(x)
    x = x[~np.isnan(x)]
    z = (x - x.mean()) / x.std(ddof=1)
    stat, p = kstest(z, 'norm')
    print(f"KS normality — {label} : D={stat:.3f} | p={p:.3g}")
    return stat, p

# 1) KS sur la variable globale
ks_normality_report(client_stats["panier_moyen"], "panier_moyen (global)")

# 2) QQ-plot global (visuel “cloche”)
fig, ax = plt.subplots(figsize=(6,6))
probplot(client_stats["panier_moyen"].dropna(), dist="norm", plot=ax)
ax.set_title("QQ-plot — panier_moyen (global)")
ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# age_group
bins = [10, 20, 30, 40, 50, 60, 100]
labels = ["10-19", "20-29", "30-39", "40-49", "50-59", "60+"]
client_stats["age_group"] = pd.cut(client_stats["age"], bins=bins, labels=labels, right=False)

# groups + observed=True
groups = [g["panier_moyen"].dropna().values
          for _, g in client_stats.groupby("age_group", observed=True)
          if len(g) > 0]

lev_stat, lev_p = levene(*groups, center="median")
print(f"Levene (variances) — panier_moyen ~ age_group : stat={lev_stat:.2f} | p={lev_p:.3g}")
```

```python
from scipy.stats import pearsonr, spearmanr

# Pearson/Spearman — Âge vs montant total des achats (CA total)
x = client_stats["age"]
y = client_stats["ca_total"]

r_p, p_p = pearsonr(x, y)
r_s, p_s = spearmanr(x, y)

print(f"Pearson(age vs ca_total)  r = {r_p:.3f} | p-value = {p_p:.4g}")
print(f"Spearman(age vs ca_total) ρ = {r_s:.3f} | p-value = {p_s:.4g}")

plt.style.use("dark_background")
fig, ax = plt.subplots(figsize=(15, 6))

# Nuage de points
ax.scatter(x, y, alpha=0.35)

# Droite de tendance (régression linéaire)
m, b = np.polyfit(x, y, 1)                     # slope + intercept
x_line = np.linspace(x.min(), x.max(), 200)
ax.plot(
    x_line,
    m * x_line + b,
    color="#ffcc00",
    linewidth=2.8,
    alpha=0.95,
    label="Tendance linéaire"
)

ax.set_title("Âge vs CA total (dataset sans BtoB)")
ax.set_xlabel("Âge")
ax.set_ylabel("CA total (€)")
ax.grid(alpha=0.2)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Pearson/Spearman — Âge vs fréquence d'achat (nb_transactions)
x = client_stats["age"]
y = client_stats["nb_transactions"]

r_p, p_p = pearsonr(x, y)
r_s, p_s = spearmanr(x, y)

print(f"Pearson(age vs nb_transactions)  r = {r_p:.3f} | p-value = {p_p:.4g}")
print(f"Spearman(age vs nb_transactions) ρ = {r_s:.3f} | p-value = {p_s:.4g}")

plt.style.use("dark_background")
fig, ax = plt.subplots(figsize=(15, 6))
ax.scatter(x, y, alpha=0.35)
ax.set_title("Âge vs fréquence d'achat (dataset sans BtoB)")
ax.set_xlabel("Âge")
ax.set_ylabel("Nombre de transactions")
ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# Pearson/Spearman — Âge vs taille du panier moyen (panier_moyen)
x = client_stats["age"]
y = client_stats["panier_moyen"]

r_p, p_p = pearsonr(x, y)
r_s, p_s = spearmanr(x, y)

print(f"Pearson(age vs panier_moyen)  r = {r_p:.3f} | p-value = {p_p:.4g}")
print(f"Spearman(age vs panier_moyen) ρ = {r_s:.3f} | p-value = {p_s:.4g}")

plt.style.use("dark_background")
fig, ax = plt.subplots(figsize=(15, 6))
ax.scatter(x, y, alpha=0.35)
ax.set_title("Âge vs panier moyen (dataset sans BtoB)")
ax.set_xlabel("Âge")
ax.set_ylabel("Panier moyen (€)")
ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
