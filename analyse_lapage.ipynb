# 🎯 Objectif du projet — Lapage 📚

Ce notebook a pour objectif d’**analyser les ventes et le comportement des clients de l’entreprise Lapage**, une librairie ayant développé son activité de vente en ligne, afin d’en extraire des **insights clairs et exploitables pour le comité de direction (CODIR)**.

L’analyse vise à fournir une **vision synthétique et stratégique de la performance commerciale**, ainsi qu’une meilleure compréhension des profils et habitudes des clients.

---

## 🧩 Contexte

Initialement positionnée comme une **librairie physique**, l’entreprise **Lapage** a lancé son site de vente en ligne il y a deux ans afin de répondre à la demande croissante de ses clients et d’élargir son activité.

Dans ce contexte, la direction souhaite aujourd’hui :

- mieux **suivre l’évolution de l’activité e-commerce** ;
- comprendre les **dynamiques de vente** ;
- analyser le **comportement des clients en ligne**.

Pour mener cette étude, trois jeux de données sont mis à disposition :

- un fichier **clients** (genre, année de naissance) ;
- un fichier **produits** (prix et catégories) ;
- un fichier **transactions** (dates d’achat, clients et produits).

---

## 📈 Objectifs principaux

Le travail se structure autour de **deux axes complémentaires**.

### Axe 1 — Analyse de l’activité commerciale

- Étudier l’**évolution du chiffre d’affaires dans le temps** ;
- Mettre en évidence les tendances à l’aide de **moyennes mobiles** ;
- Analyser le **nombre de clients**, de **transactions** et de **produits vendus** ;
- Identifier les **produits les plus performants (tops)** et les **moins performants (flops)** ;
- Étudier la **répartition des ventes par catégorie**.

---

### Axe 2 — Analyse du comportement client

- Étudier la **répartition du chiffre d’affaires par client** ;
- Analyser la **dispersion des ventes** à l’aide d’une **courbe de Lorenz** ;
- Mettre en évidence des **corrélations entre les variables clients** :
  - âge ;
  - genre ;
  - fréquence d’achat ;
  - panier moyen ;
  - catégories de livres achetées.
- Appliquer et justifier l’utilisation de **tests statistiques adaptés** :
  - test du **χ² (khi-deux)** ;
  - **ANOVA** ou **test de Student** ;
  - **corrélation de Pearson ou de Spearman**.
- Identifier et traiter les **clients atypiques (BtoB)** afin de garantir la pertinence des analyses.

---

## 🧠 Finalité

Les résultats issus de cette étude permettront :

- d’**accompagner la prise de décision stratégique du CODIR** ;
- d’orienter les **choix marketing et commerciaux** (offres, ciblage, positionnement) ;
- de proposer une **lecture claire, pédagogique et exploitable des données**, adaptée à un public non technique.

---

> 💡 *Cette analyse a été menée dans une démarche de rigueur, de clarté et de reproductibilité, afin de fournir à Lapage une vision fiable et structurée de son activité e-commerce et du comportement de ses clients.*

```python
cols_tx = ["id_prod", "date", "session_id", "client_id"]
rows = len(transactions)
nulls = transactions[cols_tx].isna().sum()

summary_missing = pd.DataFrame({
    "Non-null": rows - nulls,
    "Valeurs manquantes": nulls,
    "% manquantes": (nulls / rows * 100).round(2)
})

display(summary_missing)

# Lignes "exploitables" = toutes les colonnes clés présentes
tx_ok = transactions.dropna(subset=cols_tx).copy()

# Lignes "incomplètes" = au moins 1 colonne clé manquante
tx_missing = transactions[transactions[cols_tx].isna().any(axis=1)].copy()

# Cas très fréquent : lignes "totalement vides" (toutes NaN sur les 4 colonnes)
tx_all_nan = transactions[transactions[cols_tx].isna().all(axis=1)].copy()

print("Nb lignes total :", rows)
print("Nb lignes exploitables (4 champs présents) :", len(tx_ok))
print("Nb lignes incomplètes (>=1 champ manquant) :", len(tx_missing))
print("Nb lignes entièrement vides (4 champs manquants) :", len(tx_all_nan))
```

```python
# Contrôle 1 — Détection des doublons techniques (Transactions exploitables)

dup_exact_ok = tx_ok.duplicated().sum()
dup_strict_ok = tx_ok.duplicated(subset=cols_tx).sum()

print("\nDoublons exacts (ligne complète) sur lignes exploitables :", dup_exact_ok)
print("Doublons stricts sur (client, session, produit, date) sur lignes exploitables :", dup_strict_ok)

if dup_exact_ok > 0:
    display(tx_ok[tx_ok.duplicated(keep=False)].sort_values(cols_tx).head(20))
```

```python
# Contrôle 2 — Répétitions métier (achats multiples dans une même session)

tx_repeat = (
    tx_ok.groupby(["client_id", "session_id", "id_prod"])
         .size()
         .reset_index(name="nb_occurrences")
)

triplets = len(tx_repeat)
rep_gt1 = (tx_repeat["nb_occurrences"] > 1).sum()
part_rep = (rep_gt1 / triplets * 100) if triplets else 0
max_rep = tx_repeat["nb_occurrences"].max() if triplets else 0

print("\nTriplets analysés (client, session, produit) :", triplets)
print("Répétitions (nb_occurrences > 1) :", rep_gt1)
print("Part des répétitions : {:.2f}%".format(part_rep))
print("Maximum observé (même produit, même session) :", max_rep)

# Aperçu des cas les + élevés
display(tx_repeat.sort_values("nb_occurrences", ascending=False).head(10))
```
